```python
import pandas as pd
import altair as alt
# This library is essential in Colab/Jupyter to explicitly render multiple charts in one cell
from IPython.display import display

# --- 1. DATA DEFINITION ---
# Data for 3-Year Total Growth (Approx. Nov 2022 to Nov 2025)
data_3yr = {
    'Ticker': ['GOOGL', 'META', 'MSFT', 'AMZN', 'UBER', 'CRM', 'S&P 500'],
    'Growth_Pct': [223.3, 600.0, 132.5, 178.2, 227.5, 84.7, 95.03]
}
df_3yr = pd.DataFrame(data_3yr)

# Data for 5-Year Total Growth (Approx. Nov 2020 to Nov 2025)
data_5yr = {
    'Ticker': ['GOOGL', 'META', 'MSFT', 'AMZN', 'UBER', 'CRM', 'S&P 500'],
    'Growth_Pct': [300.9, 396.6, 269.1, 142.0, 310.2, 107.0, 114.28]
}
df_5yr = pd.DataFrame(data_5yr)


# --- 2. CHART GENERATION FUNCTION ---
def create_growth_chart(df: pd.DataFrame, years: int, bar_color: str):
    """Generates an interactive Altair bar chart for growth comparison."""

    # Find the benchmark value to draw the line
    benchmark_value = df[df['Ticker'] == 'S&P 500']['Growth_Pct'].iloc[0]

    # Create the bar chart
    chart_bars = alt.Chart(df).mark_bar().encode(
        # Sort by Growth_Pct in descending order
        x=alt.X('Ticker', sort=alt.EncodingSortField(field='Growth_Pct', op='sum', order='descending'), title='Ticker / Benchmark'),
        y=alt.Y('Growth_Pct', title=f'{years}-Year Total Growth (%)'),
        color=alt.condition(
            alt.datum.Ticker == 'S&P 500',  # Highlight the S&P 500 bar in grey
            alt.value('grey'),
            alt.value(bar_color) # Use different colors for visual distinction
        ),
        tooltip=[
            'Ticker',
            alt.Tooltip('Growth_Pct', title=f'{years}-Yr Growth %', format='.1f')
        ]
    ).properties(
        title=f'{years}-Year Total Growth Comparison (Stock vs. S&P 500)'
    )

    # Create the horizontal benchmark rule (dashed red line)
    chart_rule = alt.Chart(pd.DataFrame({'y': [benchmark_value]})).mark_rule(
        color='red',
        strokeDash=[5, 5],
        size=1.5
    ).encode(
        y='y'
    )

    # Combine the bars and the rule, and make the chart interactive
    return (chart_bars + chart_rule).interactive()


# --- 3. DISPLAY & SAVE CHARTS ---

# 3-Year Chart
chart_3yr = create_growth_chart(df_3yr, 3, 'darkgreen')
# Saves the chart as an HTML file for portability (e.g., GitHub, reports)
chart_3yr.save('3_year_growth_chart.html')
display(chart_3yr) # Explicitly displays the first chart

# 5-Year Chart
chart_5yr = create_growth_chart(df_5yr, 5, 'darkblue')
# Saves the chart as an HTML file for portability
chart_5yr.save('5_year_growth_chart.html')
display(chart_5yr) # Explicitly displays the second chart
```
